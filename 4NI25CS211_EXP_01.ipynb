```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pit
import seaborn as sns
data = {'name':['mokshagna','mohammad teja','Amit',"Luffy","Zoro","Sanj"],
         'marks':[100,100,93,60,100,50],
          'hours_studied':[4,5,3,1,2,3]}
df=pd.DataFrame(data)
print(df)
```

```python
sns.barplot(x='name',y='hours_studied',data=df)
pit.show()
```
